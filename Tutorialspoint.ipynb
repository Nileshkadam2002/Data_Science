# Web Mapping With Python—Interactive Mapping Of Population And Volcanoes

```python
import folium
```

```python
def get_marker_color(index):
    colors = ['red', 'blue', 'green', 'purple', 'orange', 'darkred', 'lightred', 'beige', 'darkblue', 'darkgreen', 'cadetblue', 'darkpurple', 'white', 'pink', 'lightblue', 'lightgreen', 'gray', 'black', 'lightgray']
    return colors[index % len(colors)]
```

Now I will update the existing for-loop to use this color generation function.

```python
map=folium.Map(location=[80,-100],zoom_start=6)
```

```python
coordinates = [[38.5, -99.0], [39.0, -98.0], [40.0, -101.0]]
```

```python
data = [
    {"location": [38.5, -99.0], "name": "Location A"},
    {"location": [39.0, -98.0], "name": "Location B"},
    {"location": [40.0, -101.0], "name": "Location C"}
]
```

```python
map = folium.Map(location=[80,-100],zoom_start=6)

for i, item in enumerate(data):
    folium.Marker(
        location=item["location"],
        popup=item["name"],
        icon=folium.Icon(color=get_marker_color(i))
    ).add_to(map)
```

```python
map.save("Map2.html")
```

```python
for i, coordinate in enumerate(coordinates):
    folium.Marker(
        location=coordinate,
        popup="Hi, I am a marker!",
        icon=folium.Icon(color=get_marker_color(i))
    ).add_to(map)
```
