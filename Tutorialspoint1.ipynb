### Convert Image to Different Color Spaces

```python
img_bgr = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
print("Shape after converting to BGR:", img_bgr.shape)

img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
print("Shape after converting to HSV:", img_hsv.shape)
cv2_imshow(img_hsv)
```

Here, I first converted the grayscale image to BGR (Blue, Green, Red) format, as `cv2.cvtColor` typically expects BGR input for conversion to HSV. Then, I converted the BGR image to HSV (Hue, Saturation, Value) color space and displayed it. The HSV color space is often used in image processing for tasks like color-based object tracking or segmentation.
